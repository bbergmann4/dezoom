## Q2 

Payment Types:  First exploring payment types, then getting a percentage
Result:  26.66% of rows have payment type credit card

```python
r = con.execute("SELECT payment_type, count(*) as rows, sum(total_amt) as total_fare FROM taxi_pipeline_duckdb_dataset.taxi_data GROUP BY 1")
print(r.fetchdf())
```

```python
r = con.execute("SELECT SUM(CASE WHEN payment_type = 'Credit' THEN 1.0 ELSE 0.0 END)/SUM(1.0) FROM taxi_pipeline_duckdb_dataset.taxi_data LIMIT 5")
print(r.fetchdf())
```
